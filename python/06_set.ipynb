# 세트(Set)

- 수학의 집합개념을 구현한 데이터타입
- 데이터에 순서가 없고, 중복해서 사용이 불가
- 교집합, 합집합, 차집합을 구하는 메서드

## 1. 세트 생성

```python
set1 = {1, 2, 3}
set1
```

```python
fruits_set = {'Apple', 'Lemon', 'Apple'}    # 데이터 중복 불가
fruits_set
```

## 2. 세트(집합) 연산
* 교집합
  - set객체.intersection(set객체)
  - &
    
* 합집합
  - set객체.union(set객체)
  - |
    
* 차집합
  - set객체.difference(set객체)

```python
set2 = {2, 3, 4}
print(set1, set2)
```

```python
# 교집합
set1.intersection(set2)
set1 & set2
```

```python
# 합집합
set1.union(set2)
set1 | set2
```

```python
# 차집합
set1.difference(set2)
```

## 4. 아이템 추가

- set객체.add(아이템)
- set객체.update(set객체)

```python
# fruits_set에 'Tomato' 추가
fruits_set.add('Tomato')
```

```python
fruits_set
```

```python
vegetable_set = {'carrot', 'onion'}
vegetable_set
```

```python
# fruits_set에 vegetable_set 추가하여 fruits_set이 update됨
fruits_set.update(vegetable_set)
```

```python
fruits_set
```

- set객체.clear() : 아이템 전체 삭제

```python
fruits_set.remove('carrot')
```
