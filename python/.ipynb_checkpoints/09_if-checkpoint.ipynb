# if문

```python
# 변수 x의 값이 10보다 크거나 같으면 'pass'를 출력

x = 11
if x >= 10 :
    print('pass')
```

```python
temp = 12

if temp > 0 :
    print('아아')
else :
    print('뜨아')
```

```python
today_temp = 25

if today_temp <= 0 :
    if today_temp == 0 :
        print('미지근')
    else :
        print('뜨')
else :
    print('아')
```

# 반복문

```python
n_list1 = [10, 20]
n_list2 = [30, 40]

for x in n_list1 :
    for y in n_list2 :
        print(f"x: {x}, y: {y}")
```

```python
n_list1 = [10, 20]
n_list2 = [30, 40]

for i1, x in enumerate(n_list1) :
    for y in n_list2 :
        print(f"x: {x}, y: {y}", i1)
```

```python
# 구구단 3단 출력

for j in range(1,10) :
    print(f"[{j}단]")
    for i in range(1,10) :
        print(f"{j} x {i} = {j*i}", end='\t')
    print("\n")
```
