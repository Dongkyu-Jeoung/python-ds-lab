# Module(모듈)

- 코드와 함수가 저장된 .py 파일
- 파이썬(.py) 파일로 생성된 모듈은 import 통해 사용
- 장점
  - 생성된 코드의 재사용성을 높임
  - 공동 작업시 유리함
  - 관리가 수월함
- 모듈 생성
  - 모드 이름 '모듈이름.py'로 저장
  - 주피터 노트북 사용시 내장함수 '%%writepfile' 코드를 사용함

```python
%%writefile fmodule.py

def add(a, b) :
    print(a+b)
```

```python
import fmodule as fm

fm.add(1, 5)
```
