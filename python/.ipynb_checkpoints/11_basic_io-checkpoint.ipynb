# 화면 출력

- print() 함수를 이용하여 화면 출력
- 기본출력 : 출력 형식 지정하지 않고 데이터
- 형식(attribute) 지정 : sep, end, .. 속성을 적용하여 다양하게 출력

```python
print('hello python')
```

```python
# 이름은 '홍길동'이고 나이는 30살 입니다.
name = "홍길동"
age = 30
print(f"이름은 '{name}'이고 나이는 {age}살 입니다.")
```

# 키보드 입력

- 키보드로 데이터를 입력받기 위해서는 input() 함수 사용
- 형식 : 변수 = input('문자열')

```python
name = input('이름을 입력해주세요 : ')
print(name)
```

```python
# 입력받은 단을 통해 구구단 출력

dan = input('몇 단?')

print(f"[{dan}단]")
for i in range(1,10) :
    print(f"{dan} x {i} = {int(dan)*i}")        # input을 통해 받은 숫자는 str으로 저장됨 (캐스팅 필요)
```
