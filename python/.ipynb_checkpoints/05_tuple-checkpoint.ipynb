# 튜플(Tuple)

- 리스트와 유사
- 한 번 입력 한 후에 항목을 ⭐변경할 수 없는⭐ 특징
- 소괄호()를 사용해서 생성 또는 괄호 없이 생성
- 튜플의 항목은 숫자, 문자열, 불린, 리스트, 튜플, 세트, 딕셔너리 등으로 만들 수 있음

## 1. 튜플 생성

```python
tuple1 = (1, 2, 3)
tuple1
```

```python
tuple1[0:2]
```

## 2. 튜플 아이템 선택

```python
fruits_tuple = ('Apple', 'Tomato')
fruits_tuple
```

```python
fruits_tuple[:]
```

## 3. 튜플 -> 리스트로 형변환

```python
fruits_tuple = list(fruits_tuple)
type(fruits_tuple)
```

```python
fruits_tuple.append('orange')
fruits_tuple
```
