# Print()

## 1. 기본 문법
- print(value, ..., sep='구분기호', end='\n', flush=False)

```python
print(1, 2, 3, sep='-')
```

```python
name = '홍길동'
age = 30
```

```python
# 4.format{} 포맷
print("이름: {}, 나이: {}". format(name, age))
```

```python
# 5. f-string (권장)
print(f"이름: {name}, 나이: {age}")
```

```python
# 줄바꿈 없이 옆으로 출력
for i in range(5) :           # 0, 1, 2, 3, 4
    print(i, end='\t')
```
