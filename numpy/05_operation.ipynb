# 배열 연산
- 배열의 shape이 같으면 데이터별로 연산이 진행됨

```python
import numpy as np
```

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

print(arr1, arr2)
```

```python
print(arr1 + arr2)
print(arr1 - arr2)
print(arr1 * arr2)
print(arr1 / arr2)
```

- 스칼라(단일 숫자)와 연산

```python
print(arr1 + 10)
print(arr1 - 10)
print(arr1 * 10)
print(arr1 / 10)
```

- 2차원 배열 연산

```python
arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])

print(arr1.shape, arr2.shape)
```

```python
print(arr1 + arr2)
print(arr1 - arr2)
print(arr1 * arr2)
print(arr1 / arr2)
```

## 정렬 - sort() 와 argsort()

### (1) sort()
- numpy.sort() vs numpy.ndarray.sort() : 원본 배열의 유지 여부가 핵심
- 원본 유지 O : numpy.sort()
- 원본 유지 X : numpy.ndarray.sort()

```python
org_arr = np.array([2, 1, 3, 6, 9])
org_arr
```

```python
# 오름 차순 정렬 후 새 배열 반환
sort_arr = np.sort(org_arr)
sort_arr
```

```python
print(org_arr, sort_arr)
```

```python
# 원본 배열을 직접 오름차순 정렬
org_arr.sort()
org_arr
```

- 내림차순 정렬 : 배열[ : : -1]

```python
org_arr = np.array([8, 4, 9, 2, 5])
sort_arr = np.sort(org_arr)[::-1]
sort_arr
```

```python
org_arr = np.array([8, 4, 9, 2, 5])
org_arr.sort()
org_arr[::-1]
org_arr
```

- 2차원 배열 정렬

```python
arr = np.array([[8, 12],
               [7, 1]])
arr
```

```python
# axis = 0 : row 기준 정렬(세로축 비교) / axis = 1 : col 기준 정렬(가로축 비교)
sort_arr1 = np.sort(arr, axis = 0)
sort_arr2 = np.sort(arr, axis = 1)
print(sort_arr1, sort_arr2, sep='\n')
```

### (2) argsort()

- 정렬된 인덱스 반환

```python
names = np.array(['홍길동', '김유신', '강감찬', '이순신', '스미스'])
score = np.array([72, 81, 95, 66, 88])
```

```python
# 성적순으로 names 배열을 오름차순 정렬
idx = np.argsort(score)
target = []

for i in idx:
    target.append(str(names[i]))

target
```

```python
names[idx]        # 팬시 인덱싱
```

```python
# 성적순으로 names 배열을 내림차순 정렬
idx = np.argsort(score)[::-1]
names[idx]
```

# 집계 함수

```python
arr = np.arange(1, 10)
arr
```

```python
print(np.sum(arr))    # 합계
print(np.mean(arr))   # 평균
print(np.min(arr))    # 최소
print(np.max(arr))    # 최대
print(np.std(arr))    # 표준편차
print(np.var(arr))    # 분산
print(np.median(arr))    # 중앙값
```
