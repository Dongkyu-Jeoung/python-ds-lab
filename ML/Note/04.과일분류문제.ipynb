#### 머신러닝을 위한 데이터 준비

- feature data와 target data 분리

```python
# feature
fruitsData = fruits[['length', 'weight']]
fruitsData.head()
```

```python
# target
fruitsTarget = fruits['name']
fruitsTarget.head()
```

### 최근접 이웃 (kNN : k Nearest Neighbors) 분류 알고리즘

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
# 모델 생성
kn = KNeighborsClassifier()     # default 5, hyper parameter 조정 : n_neighbors
```

```python
# 학습시키기
kn.fit(fruitsData, fruitsTarget)
```

```python
# 정확도
kn.score(fruitsData, fruitsTarget)
```

```python
# 예측해보기
# length : 30, weight : 600
kn.predict([[30, 600], [10, 200]])
```
