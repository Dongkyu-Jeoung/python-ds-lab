#### sckit learn의 Machine Learning 종류

```python
import pandas as pd
```

```python
csv = pd.read_csv('../Data/iris.csv')
csv.head()
```

```python
# feature와 target
iris_data = csv.iloc[:, 0:-1]
iris_label = csv['Name']
```

```python
# Train과 Test
from sklearn.model_selection import train_test_split

train_input, test_input, train_target, test_target = \
    train_test_split(
        iris_data,
        iris_label,
        random_state= 42,
        stratify= iris_label
    )

train_input.info()
```

#### kNN

```python
from sklearn.neighbors import KNeighborsClassifier

knn_clf = KNeighborsClassifier()
knn_clf.fit(train_input, train_target)
print(f'KNN : {knn_clf.score(test_input, test_target)}')
```

#### SVM

```python
from sklearn.svm import SVC

svm_clf = SVC()
svm_clf.fit(train_input, train_target)
print(f'SVM : {svm_clf.score(test_input, test_target)}')
```

#### Gaussian Process

```python
from sklearn.gaussian_process import GaussianProcessClassifier

gp_clf = GaussianProcessClassifier()
gp_clf.fit(train_input, train_target)
print(f'Gaussian Process : {gp_clf.score(test_input, test_target)}')
```

#### Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier

dt_clf = DecisionTreeClassifier()
dt_clf.fit(train_input, train_target)
print(f'Deicision Tree : {dt_clf.score(test_input, test_target)}')
```

#### Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

rf_clf = RandomForestClassifier()
rf_clf.fit(train_input, train_target)
print(f'Random Forest : {rf_clf.score(test_input, test_target)}')
```

#### AdaBoost

```python
from sklearn.ensemble import AdaBoostClassifier

ada_clf = AdaBoostClassifier()
ada_clf.fit(train_input, train_target)
print(f'Ada Boost : {ada_clf.score(test_input, test_target)}')
```

#### Neural Net

```python
from sklearn.neural_network import MLPClassifier

mlp_clf = MLPClassifier()
mlp_clf.fit(train_input, train_target)
print(f'Neural Net : {mlp_clf.score(test_input, test_target)}')
```

```python
from sklearn.discriminant_analysis import QuadraticDiscriminantAnalysis

qda_clf = QuadraticDiscriminantAnalysis()
qda_clf.fit(train_input, train_target)
print(f'QuadraticDiscriminantAnalysis : {qda_clf.score(test_input, test_target)}')
```

#### Naive Bayes

```python
from sklearn.naive_bayes import GaussianNB

nb_clf = GaussianNB()
nb_clf.fit(train_input, train_target)
print(f'Naive Bayes : {nb_clf.score(test_input, test_target)}')
```
