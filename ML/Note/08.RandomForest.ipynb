```python
import pandas as pd
```

```python
iris = pd.read_csv('../Data/iris.csv')
iris.head()
```

```python
clf.score(test_input, test_target)
```

#### 랜덤포레스트를 디용한 버섯 분류

```python
import pandas as pd
```

```python
mr = pd.read_csv('../Data/mushroom.csv', header= None)     # header를 지정 안해주면 첫 줄이 column명으로 들어감
mr.head()
```

- 한 줄이 버섯 한 종류
- 첫번째 열 : p(독버섯), e(식용버섯)        -> 얘가 target임
- 두번째 열 : 버섯의 머리 모양 -> b(종), c(원뿔), x(볼록), f(평평), k(혹), s(오목)
- 네번째 열 : 버섯의 머리 색깔 -> n(갈색), b(황갈색), c(연한갈색)

#### Feature Data가 알파벳 1자리 문자 -> 숫자로 변환

----
#### 연습

```python
# 알파벳 -> ASCII
ord('x')
```

```python
# ASCII -> 알파벳
chr(120)
```

```python
# mr의 1번 컬럼을 숫자로 변경
[ord(i) for i in mr.loc[:, 1]]
```

----

```python
target = []
data = []

for row_index, row in mr.iterrows() :
    target.append(row.loc[0])
    row_data = []
    for v in row.loc[1:]:
        row_data.append(ord(v))
    data.append(row_data)
```

```python
data[:2]
```

```python
print(target[:2])
```

#### DataFrame 만들기

```python
# Target을 기준으로 DataFrame 만들기
targetTemp = pd.DataFrame(target)
targetTemp.head()
```

```python
# data를 dataframe으로 만들고 column 이름을 겹치지 않게 하기
dataTemp = pd.DataFrame(data)
dataTemp.rename(columns= lambda x : x + 1, inplace = True)
dataTemp.head()
```

```python
# 2개의 dataframe 합치기
mr2 = pd.concat([targetTemp, dataTemp], axis='columns')
mr2.head()
```

```python
mr2.info()
```

#### RandomForest로 분류하기

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
```

```python
# data / target 분리
mr_data = mr2.iloc[:,1:]
mr_target = mr2[0]

print(mr_target)
```

```python
train_input, test_input, train_target, test_target = \
    train_test_split(
        mr_data,
        mr_target,
        random_state= 42,
        stratify= mr_target
    )
```

```python

# 모델 생성
clf = RandomForestClassifier()

# 모델 학습
clf.fit(train_input, train_target)

# 검증
clf.score(test_input, test_target)
```

```python
clf.get_params()
```
