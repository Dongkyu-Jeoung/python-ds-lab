## 머신러닝 이란?
: 수 많은 데이터를 학습시켜 패턴을 찾아내고 데이터를 분류(Classification) 하거나 예측(Predict)    
- 일반 소프트웨어(결정론적)는 언제나 똑같은 과정을 거쳐 똑같은 결과
- 머신러닝 소프트웨어(비결정론적)는 데이터와 알고리즘에 따라 다른 결과를 추론

## 작업 방법
- 규모가 있는 샘플 데이터를 입력으로 넣어 분석 시킴
- 데이터에서 일정한 규칙 발견
- 찾아낸 규칙을 기반으로 데이터를 분류, 예측

## 어떻게 특징(feature)과 규칙(rule)을 찾을까?
- 특정 추출 데이터가 어떤 특징을 가지고 있는지 찾고 벡터로 만드는 처리과정
- 어떤 특징을 추출할지는 분석가가 결정
- 회귀분석(Regression) : 데이터가 연속된 값일 때 사용 / y = ax + b
    - y : 종속 변수(Target)
    - x : 독립 변수(Feature)

## 머신러닝의 종류
#### 1. 지도학습(Supervised Learning)
: 학습 시 데이터와 답을 같이 학습 후 다른 데이터의 답을 예측

#### 2. 비지도학습(Unsupervised Learning)
: 학습 시 데이터만 학습 후 다른 데이터의 규칙성을 찾는다
- 군집 분석(Cluster Analysis)
- 주성분 분석(Principal Component Analysis, PCA)

#### 3. 강화학습(Reinforcement Learning)
: 부분적으로 답을 입력한 후 데이터를 기반으로 최적의 답을 찾아냄

## 머신러닝의 흐름
(1) 데이터 수집    
(2) 데이터 가공 : 결측치, 중복값 처리    
(3) 데이터 학습    
    (3-1) 학습 방법 선택 (알고리즘)    
    (3-2) 매개변수(Hyper Parameter) 조정
    (3-3) 모델 학습    
(4) 모델 평가    
(5) 예측력과 오분류표가 좋은 경우 다음단계 아니면 (3)부터 재실행    

## 머신러닝의 응용 분야

#### 분류(Classification)
- 데이터에 label을 분류할 수 있음
- 스팸메일 분류, 필기 인식, 증권 사기

##### 이진 분류
: 둘 중 하나의 값으로 분류

##### 다중 분류
: 여러개의 분류 값 중 하나의 값으로 예측

#### 군집(Clustering)
: 값의 유사성 기반으로 데이터를 여러 그룹으로 나눔
- 광고

#### 추천(Recommendation)
: 특정 데이터를 기반으로 다른 데이터 추천

#### 차원 축소(Dimensional Reduction)
: 데이터의 특성을 유지하면서 데이터의 양을 줄임

## 머신러닝의 필수 개념

#### 과대적합(Overfitting)
: 학습데이터 외의 데이터에는 정확도가 낮게 나옴

#### 과소적합(Underfitting)
: 학습데이터보다 일반데이터가 정확도가 높게 나옴
