- svm(svm_iris.h5)

저장된 모델을 불러서 예측해보기
- SepalLength : 5.1
- SepalWidth : 3.0
- PetalLength : 1.3
- PetalWidth : 0.2

```python
# 데이터 가져오기
iris_df = pd.read_csv('../Data/iris.csv')
iris_df.head()
```

```python
# feature / target 분리
iris_data = iris_df.iloc[:, 0:-1]
iris_target = iris_df['Name']
```

```python
# train / test dataset 구성
from sklearn.model_selection import train_test_split

train_input, test_input, train_target, test_target = \
    train_test_split(
        iris_data,
        iris_target,
        random_state= 42,
        stratify= iris_target
    )

train_input.info()
```

```python
# 모델 생성 (SVM, RF, kNN)
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

clf_svm = SVC()
clf_rf = RandomForestClassifier()
clf_knn = KNeighborsClassifier()

models = [clf_svm, clf_rf, clf_knn]
score = {}

for model in models :
    model.fit(train_input, train_target)
    score[model] = model.score(test_input, test_target)

print(score)
```

```python
# 에측 데이터
data = {
    'SepalLength' : 5.1,
    'SepalWidth' : 3.0,
    'PetalLength' : 1.3,
    'PetalWidth' : 0.2
}

new = pd.DataFrame([data])
new
```

```python
# 새로운 데이터 예측

for model in models:
    print(f"{model}, prediction : {model.predict(new)}")
```
