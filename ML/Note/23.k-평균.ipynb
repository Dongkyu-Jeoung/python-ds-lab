#### k-Means
- 대표적인 군집 알고리즘으로 군집의 중심(평균)을 찾는 작업
- 군집의 중심(평균)
- 클러스터(cluster) : 군집
- 클러스터링 : 군집화 하는 작업

```python
import numpy as np
```

```python
fruits = np.load("../Data/fruits.npy")
fruits.shape
```

```python
fruits.mean()
```

```python
# 3차원 데이터를 2차원 데이터로 변경

fruits_2d = fruits.reshape(-1, 100*100)
fruits_2d.shape
```

#### k-Means Model

```python
from sklearn.cluster import KMeans
```

```python
km = KMeans(
   n_clusters=3, # 데이터를 3종류로 만들어라
)

km.fit(fruits_2d)
```

```python
# 군집화된 Label 출력
km.labels_
```

----
#### 첫번쨰 클러스터 확인하기

```python
import matplotlib.pyplot as plt
```

```python
def draw_fruits(arr, ratio=1): # ratio : 해상도
   n = len(arr) # n : 샘플갯수
   # 한 줄에 10개씩 이미지를 그린다.
   rows = int(np.ceil(n/10)) # 올림처리
   # 행이 1개이면 열 갯수가 샘플갯수이며, 그렇지 않으면 10개로 만들어야 된다.
   cols = n if rows < 2 else 10
   fig, axs = plt.subplots(
      rows,
      cols,
      figsize = (cols*ratio, rows*ratio),
      squeeze=False
   )

   for i in range(rows):
      for j in range(cols):
         if i*10+j < n:
            axs[i,j].imshow(arr[i*10+j], cmap='gray')
         axs[i,j].axis('off')
         
   plt.show()
```

```python
draw_fruits(fruits[km.labels_ == 0]) # 0번 가장 마지막이기에 바나나가 나옴
```

```python
draw_fruits(fruits[km.labels_ == 1])
```

```python
draw_fruits(fruits[km.labels_ == 2])
```

#### 클러스터의 중심

```python
km.cluster_centers_
# 이미지 3개의 중심점
```

```python
draw_fruits(km.cluster_centers_.reshape(-1,100,100),ratio=3)
```

#### 최적의 k 찾기

##### Elbow Method (최적의 군집수 구하기) : 데이터가 급격히 변하는 구점




```python
inertia = []
for k in range(2,7):
   km = KMeans(
      n_clusters=k
   )
   km.fit(fruits_2d)
   inertia.append(km.inertia_)

plt.plot(
   range(2,7),
   inertia
)

plt.show()
```

```python
inertia = []
for k in range(2,6+1):
   km = KMeans(
      n_clusters=k
   )
   km.fit(fruits_2d)
   inertia.append(km.inertia_)

plt.plot(
   range(2,6+1),
   inertia
)

plt.show()
```
