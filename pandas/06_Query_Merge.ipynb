## 1. Query를 이용한 데이터 조회
- query() : 조건식을 문자열 형식으로 작성하여 조회

```python
# exam.csv 데이터 로드
exam_df = pd.read_csv('data/exam.csv')
exam_df.head()
```

```python
# exam_df의 형태, 정보 확인
print(f"shape : {exam_df.shape}, info : {exam_df.info()}")
```

```python
# 수학 점수가 70점 이상인 데이터 조회
exam_df.query('math >= 70')
```

```python
# 수학 점수가 70점 이상인 2반 학생들 데이터 조회
# 두 개 이상의 조건은 And(&), OR(|) 연산 적용
query_exp = '(math >= 70) & (nclass == 2)'
exam_df.query(query_exp)
```

```python
# 1반 이거나 영어 점수가 90점 이상인 데이터 조회
query_exp = 'nclass == 1 | english >= 90'
exam_df.query(query_exp)
```

```python
# nclass 기준 groupby
exam_df.groupby(by = 'nclass').count()

# 1, 3, 5반 데이터만 조회
query_exp = 'nclass == 1 | nclass == 3 | nclass == 5'
exam_df.query(query_exp)
query_exp2 = 'nclass in (1, 3, 5)'
exam_df.query(query_exp2)
```

```python
# 1반의 수학, 과학 점수 평균 구하기
query_exp = 'nclass == 1'
exam_df.query(query_exp)[['math', 'science']].mean()
```

```python
# 변수를 이용하여 query 조건 생성
nclass_input = int(input())
query_exp = 'nclass == @nclass_input'
exam_df.query(query_exp)
```

## 2. Merge
- SQL의 조인과 유사함
- inner join
- left join
- right join
- full outer join

### (1) Inner Join
- key를 기준으로 두 개의 데이터프레임(테이블)
- 예제) '이름'을 키로하여 Inner Join(교집합) 수행

```python
df1 = pd.DataFrame(
    {
        '이름' : ['홍길동', '이순신', '김유신'],
        '키' : [180, 170, 175]
    }
)
df1
```

```python
df2 = pd.DataFrame(
    {
        '이름' : ['홍길동', '이순신', '김영희'],
        '성별' : ['남자', '남자', '여자']
    }
)
df2
```

```python
# Inner Join, how = 'inner' 기본값, 생략 가능
pd.merge(df1, df2, how = 'inner')
```

### (2) Left Outer Join
- 왼쪽 데이터셋에 있는 데이터프레임(테이블)의 키를 기준으로 병합

```python
# df1, df2 데이터프레임 Left Outer Join(Inner Join + Left 테이블 모든 데이터셋)
pd.merge(df1, df2, how = 'left')
```

### (3) Right Outer Join
- 오른쪽 데이터프레임에 있는 모든 데이터를 병합

```python
# df1, df2 데이터프레임 Right Outer Join(Inner Join + Right 테이블 모든 데이터셋)
pd.merge(df1, df2, how = 'right')
```

### (4) Full Outer Join
- 키를 기준으로 두 데이터프레임에 존재하는 모든 데이터를 기준으로 병합

```python
# df1, df2 데이터프레임 Full Outer Join
pd.merge(df1, df2, how = 'outer')
```

## 3. Concat 함수를 이용한 데이터 병합

```python
# df1, df2를 행 기준으로 concat
pd.concat([df1, df2])
```

```python
# df1, df2를 열 기준으로 concat
pd.concat([df1, df2], axis = 1)
```
