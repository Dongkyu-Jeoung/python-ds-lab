## 2. Dataframe 생성

### (1) 리스트(list)로 생성

```python
# 파이썬의 2차원 리스트 생성
list1 = [['김철수', 28, '개발자'], ['이영희', 34, '디자이너']]

# 리스트를 데이터프레임으로 생성
list_df = pd.DataFrame(list1, columns=['이름', '나이', '직업'])
display(list_df)
```

### (2) 딕셔너리(Dict)로 생성

```python
dict = {'Name': ['이순신', '유관순','김유신','홍길동'],
        'Year': [2011, 2016, 2015, 2015],
        'Gender': ['Male', 'Female', 'Male', 'Male']
       }
dict
```

```python
# 딕셔너리로 df 생성
dict_df = pd.DataFrame(dict)
dict_df
```

```python
# dict_df 데이터프레임에 컬럼명 변경
dict_df2 = pd.DataFrame(dict)
dict_df2.columns = ['이름', '출생연도', '성별']
display(dict_df2)
```

```python
# rename 이용해서 처음부터 col명 변경
dict_df2 = pd.DataFrame(dict).rename(columns = {
    'Name' : '이름',
    'Year' : '출생연도',
    'Gender' : '성별'
})
display(dict_df2)
```
