# DataFrame Sort, Aggregation, GroupBy

````python
import warnings
warnings.filterwarnings(action='ignore')

import pandas as pd
import numpy as np
````

````python
# Titanic 데이터셋 로드
titanic_df = pd.read_csv('data/titanic/train.csv')
titanic_df.info()
````

## 1. DataFrame 정렬(Sort)

### (1) sort_values()
- DataFrame을 특성 컬럼 기준으로 정렬할 때 사용
- 문법
  
  ```python
  DataFrame.sort_values(by='컬럼리스트', ascending = True)       # default : 오름차순(asc)
  DataFrame.sort_values(by='컬럼리스트', ascending = False)      # False => 내림차순(desc)

````python
# Name 기준 오름차순 정렬
td_sort_name = titanic_df.sort_values(by = 'Name')                   # 원본 수정 X

# 정렬에 따른 인덱스 재생성
td_sort_name.reset_index(inplace = True)
td_sort_name.drop(columns='index', axis = 1, inplace = True)

td_sort_name.head()
````

````python
# Pclass 컬럼 기준으로 내림차순 정렬
td_sort_pclass = titanic_df.sort_values(by = 'Pclass', ascending = False)
td_sort_pclass.head()
````

````python
# 여러 컬럼 정렬
titanic_df.sort_values(by = ['Name', 'Pclass'])
````

## 2. 집계 함수 (Aggregation)

### (1) count()
- 결측치(NaN) 제외
- 실제 데이터 갯수 반환

````python
titanic_df.count()
````

````python
# 특정 col만 count
titanic_df['Name'].count()
````

````python
# 여러개 column 선택해서 count
titanic_df[['Name', 'Pclass', 'Fare']].count()
````
