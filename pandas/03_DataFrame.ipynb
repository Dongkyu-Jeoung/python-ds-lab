# DataFrame과 리스트, 딕셔너리, ndarray 상호변환

## 1. 1차원 데이터를 DataFrame으로 변환

```python
import warnings
warnings.filterwarnings(action='ignore')

import pandas as pd
import numpy as np
```

```python
# 1차원 리스트 생성
col_name = ['col_1']
list1 = [1, 2, 3]
print(type(list1), list1)
```

```python
# ndarray로 변환
array1 = np.array(list1)
print(type(array1), array1)
```

```python
# list1, array1를 DataFrame으로 생성(변환)
df_1 = pd.DataFrame(list1, columns = col_name, index = ['a', 'b', 'c'])
df_2 = pd.DataFrame(array1, columns = col_name)
print(df_1)
print('-' * 30)
print(df_2)
```

##  2. 2차원 데이터를 DataFrame으로 변환

```python
# 컬럼명 생성
col_names = ['col1', 'col2', 'col3']
col_names
```

```python
# 2차원 리스트 생성
list2 = [[1, 2, 3],
         [4, 5, 6]]
list2
```

```python
# list2를 ndarray로 변환
array2 = np.array(list2)
array2
```

```python
# list2, array2를 DataFrame으로 변환
df2_1 = pd.DataFrame(list2, columns = col_names)
df2_2 = pd.DataFrame(array2, columns = col_names)

display(df2_1)
print('-' * 30)
display(df2_2)
```

## 3. 딕셔너리를 DataFrame으로 변환

```python
# 딕셔너리 생성
dict1 = {
    'col1' : [1, 2, 3],
    'col2' : [4, 5, 6],
    'col3' : [7, 8, 9]
}

dict1
```

```python
# DataFrame 생성
df_dict1 = pd.DataFrame(dict1)
df_dict1
```

## 4. DataFrame을 ndarray, list, dictionary로 변환

```python
# df_dict1 데이터프레임을 ndarray로 변환
array3 = df_dict1.values
print(type(array3), array3, sep = '\n')
```

```python
# df_dict1 데이터프레임을 list로 변환
list3 = df_dict1.values.tolist()
print(type(list3), list3, sep = '\n')
```

```python
# df_dict1 데이터프레임을 dictionary로 변환
dict3 = df_dict1.to_dict()
dict3
```
