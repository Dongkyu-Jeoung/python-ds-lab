### 패션 MNIST

```python
from tensorflow import keras

(train_input, train_target), (test_input, test_target) = \
    keras.datasets.fashion_mnist.load_data()
```

```python
# train과 valid data 분리
from sklearn.model_selection import train_test_split

train_scaled = (train_input / 255.0).reshape(-1, 28*28)

train_scaled, val_scaled, train_target, val_target = \
    train_test_split(
        train_scaled,
        train_target,
        test_size= 0.2,
        random_state= 42
    )
```

#### Layer를 추가하는 방법 - 1

```python
# Layer 만들기
dense1 = keras.layers.Dense(100, activation= 'sigmoid', input_shape= (784,))        # 은닉층
dense2 = keras.layers.Dense(10, activation='softmax')                               # 출력층
```

```python
# 전체 모델에 층을 추가
model = keras.Sequential([dense1, dense2])
```

```python
# Summary
model.summary()
```

----
#### Layer 추가 방법 - 2

```python
model = keras.Sequential(
    [
        keras.layers.Dense(100, activation='sigmoid', input_shape=(784,), name='hidden'),
        keras.layers.Dense(10, activation='softmax', name='output')
    ],
    name = '패션 MNIST 모델'
)
```

```python
model.summary()
```

----
#### Layer 추가 방법 - 3

```python
model = keras.Sequential()
model.add(keras.layers.Dense(100, activation='sigmoid', input_shape=(784,)))
model.add(keras.layers.Dense(10, activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(
    loss='sparse_categorical_crossentropy',
    metrics = ['accuracy']
)

model.fit(
    train_scaled,
    train_target,
    epochs = 5
)
```

#### 입력층을 이미지 그대로 사용하기

```python
(train_input, train_target), (test_input, test_target) = \
    keras.datasets.fashion_mnist.load_data()

train_scaled = train_input / 255.0

train_scaled, val_scaled, train_target, val_target = \
    train_test_split(
        train_scaled,
        train_target,
        test_size= 0.2,
        random_state= 42
    )
```

```python
train_scaled.shape
```

```python
model = keras.Sequential()
model.add(keras.layers.Flatten(input_shape = (28, 28)))
model.add(keras.layers.Dense(100, activation='relu'))
model.add(keras.layers.Dense(10, activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(
    loss='sparse_categorical_crossentropy',
    metrics = ['accuracy']
)

model.fit(
    train_scaled,
    train_target,
    epochs = 20
)
```

```python
model.evaluate(val_scaled, val_target)
```

```python
# 모델 훈련 (epoch 조절)
# fit 만 epoch를 바꾸면서 하면 계속 누적되니까 모델 생성부터 다시 할 수 있도록 해야함
epochs_num = [5, 10, 15, 20, 50, 100]
scores = []

for epoch in epochs_num:
    model = keras.Sequential()
    model.add(keras.layers.Flatten(input_shape = (28, 28)))
    model.add(keras.layers.Dense(100, activation='relu'))
    model.add(keras.layers.Dense(10, activation='softmax'))

    model.compile(
    loss = 'sparse_categorical_crossentropy',
    metrics = ['accuracy']
    )

    model.fit(train_scaled, train_target, epochs= epoch)
    eval = model.evaluate(val_scaled, val_target)[1]
    print(f"epoch : {epoch} -> score : {eval}")
    scores.append(eval)

print(epochs_num, scores)
```
